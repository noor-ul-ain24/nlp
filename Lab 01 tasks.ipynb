```python
# Task 03: Dictionaries for Word Mapping

# (a)
word_freq = {
    "machine": 4,
    "learning": 7,
    "data": 5,
    "model": 3,
    "text": 6
}

print(word_freq)

print(word_freq["learning"])

word_freq["data"] = 9

print(word_freq)


# (b)
word_freq["neural"] = 8
word_freq["network"] = 5

del word_freq["model"]

if "text" in word_freq:
    print("The word 'text' exists in the dictionary.")
else:
    print("The word 'text' does not exist in the dictionary.")

print(word_freq)


# (c)
vocab = {
    "nlp": 0,
    "python": 1,
    "token": 2,
    "vector": 3,
    "corpus": 4
}

id_to_word = {}

for word, ID in vocab.items():
    id_to_word[ID] = word

print("Vocabulary:", vocab)
print("Reverse Vocabulary:", id_to_word)


# (d)
for word, frequency in word_freq.items():
    print(f'Word: "{word}" → Frequency: {frequency}')

print("Total number of unique words:", len(word_freq))
```

```python
# Task 04: Sets for Unique Vocabulary

# (a)
raw = ["nlp", "ai", "nlp", "deep", "learning", "ai", "nlp", "python"]

vocab = set(raw)

print("Raw list:", raw)
print("Vocabulary set:", vocab)
print("Number of unique words:", len(vocab))

# Sets automatically remove duplicate values because
# a set can contain only unique elements.


# (b)
set_A = {"machine", "learning", "data", "model"}
set_B = {"deep", "learning", "neural", "data"}

union = set_A | set_B
print("Union:", union)

intersection = set_A & set_B
print("Intersection:", intersection)

difference = set_A - set_B
print("Difference (A - B):", difference)


# (c)
stopwords = {"is", "the", "a", "in", "of", "and", "to"}

tokens = ["nlp", "is", "a", "field", "of", "ai", "and", "the", "future"]

cleaned_tokens = set(tokens) - stopwords

print("Cleaned tokens:", cleaned_tokens)


# (d)
vocab.add("transformer")
vocab.add("bert")

vocab.discard("python")

if "ai" in vocab:
    print("'ai' is still present in the vocabulary.")
else:
    print("'ai' is not present in the vocabulary.")

print("Final vocabulary:", vocab)
```

```python
# Task 06: String Cleaning Methods


# (a)
text = "   MACHINE Learning is AMAZING!!!   "

# Remove leading and trailing spaces
text = text.strip()
print("After strip():", text)

# Convert to lowercase
text = text.lower()
print("After lower():", text)

# Remove all exclamation marks
text = text.replace("!", "")
print("After replace():", text)


# (b)
# Split text into words
words = text.split()
print("Words:", words)

# Convert each word to uppercase
uppercase_words = [word.upper() for word in words]
print("Uppercase words:", uppercase_words)

# Join words using " | "
combined = " | ".join(uppercase_words)
print("Combined:", combined)


# (c)
sentence = "I love Java and Java is popular"

# Replace Java with Python
updated_sentence = sentence.replace("Java", "Python")
print("Updated sentence:", updated_sentence)

# Find first occurrence of Python
position = updated_sentence.find("Python")
print("Starting position of Python:", position)


# (d)
def normalize(text):
    text = text.strip()
    text = text.lower()
    text = text.replace(",", "")
    text = text.replace(".", "")
    
    return text


# Test the function
text1 = "   Hello, WORLD.   "
text2 = "  Natural Language, Processing is Amazing.  "

print("Cleaned text 1:", normalize(text1))
print("Cleaned text 2:", normalize(text2))
```

```python
# Task 07: String Validation Methods


# (a)
text = "NLP is the future of AI research"

if text.startswith("NLP"):
    print("The text starts with 'NLP'.")
else:
    print("The text does not start with 'NLP'.")

if text.endswith("research"):
    print("The text ends with 'research'.")
else:
    print("The text does not end with 'research'.")

position = text.find("future")
print("The word 'future' starts at index:", position)


# (b)
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(token, "→ Alphabetic (WORD)")
    elif token.isdigit():
        print(token, "→ Numeric (NUMBER)")
    else:
        print(token, "→ Mixed")


# (c)
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)


# (d)
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"


tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in tokens:
    label = validate_token(token)
    print(token, "→", label)
```

```python
# Task 08: Functions and Lambda for NLP Pipelines


# (a)
def clean(text):
    text = text.strip()
    text = text.lower()
    return text


# Equivalent lambda function
clean_lambda = lambda text: text.strip().lower()


text = "  HELLO Natural Language Processing  "

print("Regular function:", clean(text))
print("Lambda function:", clean_lambda(text))


# (b)
def tokenize(text):
    return text.split()


def remove_short(tokens, n):
    return [token for token in tokens if len(token) > n]


sentence = "I love working with NLP tools in Python daily"

tokens = tokenize(sentence)
result = remove_short(tokens, 3)

print("Tokens:", tokens)
print("After removing short words:", result)


# (c)
def word_lengths(tokens):
    return {word: len(word) for word in tokens}


tokens = ["language", "model", "ai", "processing", "deep"]

result = word_lengths(tokens)

print("Word lengths:", result)


# (d)
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered_tokens = remove_short(tokens, min_len)
    return word_lengths(filtered_tokens)


sentence1 = "  Natural Language Processing is very interesting  "
sentence2 = "  Deep Learning helps machines understand text  "

result1 = nlp_pipeline(sentence1, 3)
result2 = nlp_pipeline(sentence2, 3)

print("Pipeline Output 1:", result1)
print("Pipeline Output 2:", result2)
```

```python
# Task 09: map(), filter(), sorted() and Counter

# (a)
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# Convert words to lowercase
lowercase_words = list(map(lambda word: word.lower(), words))
print("Lowercase words:", lowercase_words)

# Calculate length of each word
word_lengths = list(map(lambda word: len(word), words))
print("Word lengths:", word_lengths)


# (b)
# Words with more than 3 characters
words_long = list(filter(lambda word: len(word) > 3, words))
print("Words longer than 3 characters:", words_long)

# Words containing the letter 't'
words_with_t = list(
    filter(lambda word: "t" in word.lower(), words)
)
print("Words containing 't':", words_with_t)


# (c)
# Alphabetical order
alphabetical = sorted(words)
print("Alphabetical:", alphabetical)

# Word length ascending
length_ascending = sorted(words, key=len)
print("Length ascending:", length_ascending)

# Word length descending
length_descending = sorted(words, key=len, reverse=True)
print("Length descending:", length_descending)


# (d)
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"

word_list = sentence.split()

freq = Counter(word_list)

print("Counter:", freq)

# 3 most common words
print("3 most common words:", freq.most_common(3))

# Frequency of python
print("Frequency of python:", freq["python"])

# Total words
print("Total number of words:", len(word_list))

# Unique vocabulary
print("Unique vocabulary size:", len(freq))
```
